```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 2, 5],
    'age':         [25, np.nan, 35, np.nan, np.nan, 40],
    'city':        ['Mumbai', 'Delhi', np.nan, 'Mumbai', 'Delhi', np.nan],
    'gender':      ['Male', 'Female', 'Female', 'Male', 'Female', 'Male'],
    'annual_income': [40000, 80000, 60000, 120000, 80000, 95000]
})
```

TASK-**1**

```python
df.isnull().sum()
```

```python
df['age']=df['age'].fillna(df['age'].median())
```

```python
df['city']=df['city'].fillna(df['city'].mode()[0])
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

TASK-**2**

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['gender']=le.fit_transform(df[['gender']])
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

en = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

city_encoded_array = en.fit_transform(df[['city']])

feature_names = en.get_feature_names_out(['city'])

df_city_encoded = pd.DataFrame(city_encoded_array, columns=feature_names, index=df.index)

# Drop the original 'city' column and concatenate the new encoded columns
df = pd.concat([df.drop('city', axis=1), df_city_encoded], axis=1)
```

```python
df
```

TASK-**3**

```python
#Scaling
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
df['annual_income']=scaler.fit_transform(df[['annual_income']])
df['age']=scaler.fit_transform(df[['age']])
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
df['annual_income']=scaler.fit_transform(df[['annual_income']])
df['age']=scaler.fit_transform(df[['age']])
```
